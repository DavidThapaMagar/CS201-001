```python
def reverse_word(word):
    stack = []
    for ch in word:
        stack.append(ch)
        output = ''
    while stack:
        output += stack.pop()
    return output
```

```python
reverse_word('mama')
```

```python
def bubble_sort(arr):
    n = len(arr)
    for i in range(n-1):
        for j in range(n-1-i):
            if arr[j]>arr[j+1]:
                arr[j],arr[j+1]=arr[j+1],arr[j]
    return arr
```

```python
arr = [1,234,24,5,3,78,24]
bubble_sort(arr)
```

```python
def traverse(head_index,storage):
    output = []
    index = head_index
    while index is not None:
        value,index = storage[index]
        output.append(value)
    return output
```

```python
traverse(0, [('A',1,),('B',2),('C',None)])
```
