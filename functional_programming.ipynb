



## Lambda Functions & Comprehensions

---

### What Are Lambda Functions? Why Do We Need Them?

In Python, a **lambda function** is a very small, anonymous function defined in a single line.  
They are useful when we need a quick, simple function — especially when passing functions into other functions like `sorted()`, `map()`, or `filter()`.

#### Why use lambdas?
- To avoid writing a full multi-line `def` block for something simple
- To make sorting, filtering, and mapping cleaner and shorter
- To create temporary functions that are used only once

Lambdas should be used **only when the function logic is simple**.

---

#### Lambda Syntax



```python
lambda arguments: expression
```



- The **arguments** go before the colon
- The **expression** is calculated and automatically returned
- **Only one expression is allowed**

This makes lambdas simple and readable.

---

### Example 1 — A Simple Square Function


```python
square = lambda num:num*num
```

```python
square(5)
```



### Explanation:
- The lambda takes the argument `x` and returns `x * x`
- This behaves exactly like a normal function
- Instead of writing a full `def square(x)`, we use a one-line lambda

Output:

### Example 2 — Add Two Numbers



```python
add = lambda a,b: a+b
add(3,7)
```



### Explanation:
- Lambdas can have multiple arguments
- The expression `a + b` is returned
- This is useful for quick operations in data processing

---

### Example 3 — Extract the First Character



```python
first = lambda s:s[0]
first("Ali")
```



### Explanation:
- Lambdas work with any data type, including strings
- This is useful when sorting or filtering based on characters

---

###  Lambda vs Def (Important)

#### Using `def`:


```python
def increment(x):
    return x + 1
```



### Using `lambda`:


```python
increment = lambda x: x+1
```

```python
increment(increment(5))
```



### Explanation:
- Both are equal in functionality
- Use `lambda` for a **1-line simple function**
- Use `def` for anything more complex

---

### When NOT to Use Lambda

Do **not** use lambda when:
- Logic takes multiple steps
- You need loops
- You want readability
- You need debugging

If logic grows → convert to a proper `def` function.

---

### Real Use Case: Sorting with Lambda

Students already know sorting. Lambdas make sorting flexible.

### Example:


```python
students = [('Sara',82),('Tom',95),('Ali',88)]
sorted(students)
```

```python
sorted(students,key = lambda t:t[1])
```

```python
students = [('Sara',82,25),('Tom',95,18),('Ali',88,20)]
```

```python
sorted(students,key = lambda t:t[2])
```



### Explanation:
- `sorted()` normally sorts by the **first element**
- `key=lambda x: x[1]` tells Python to sort by the **grade**
- This is an extremely practical use case

## LIST COMPREHENSIONS

### 🎯 What Are List Comprehensions?

A **list comprehension** lets us create lists in a clear, compact way.

Benefits:
- Shorter than loops  
- Faster than loops  
- Easier to read (for simple transformations)  
- Encourages clean “transform/filter” thinking  

---

### Basic Syntax



```python
[ expression for item in iterable ]
```



With filtering:



```python
[ expression for item in iterable if condition ]
```



---

### Example 1 — Squares of Numbers



```python
def list_squares():
    squares = []
    for num in range(10):
        squares.append(num*num)
    return squares
```

```python
list_squares()
```

```python
[num*num for num in range(10)]
```



### Explanation:
- For each number from 0 to 9, compute `x * x`
- Store all results in a list
- Equivalent to a loop but much cleaner

---

### Example 2 — Convert Temperatures (Celsius → Fahrenheit)



```python
temp_c = [0,9,20,30,45]
temp_f = [(1.8 * c) + 32 for c in temp_c]
print(temp_f)
```



### Explanation:
- Applies transformation to each item
- Perfect example of transforming each list element

---

### Example 3 — Filter Even Numbers



```python
def even_numbers():
    even_number = []
    for num in range(21):
        if num % 2 == 0:
            even_number.append(num)
    return even_number
```

```python
even_numbers()
```

```python
[num for num in range(21) if num % 2 == 0]
```



### Explanation:
- The `if` condition filters out odd numbers
- This is short for:  
  “Keep x only if x % 2 equals 0.”

---

### Example 4 — Filter Names by Length



```python
names = ['Ali','Sara','John','Tom']
[name for name in names if len(name)>3]
```

```python
names = ['Ali','Sara','John','Tom']
sorted([name for name in names if len(name)>3])
```

```python
names = [('Ali',66),('Sara',45),('John',78),('Tom',21)]
sorted([(name,grade) for (name,grade) in names if len(name)>3])
```



### Explanation:
- Filters strings based on length  
- Useful for text processing, cleaning datasets, etc.

---

### Example 5 — If/Else Inside the Expression



```python
['even' if num%2==0 else 'odd' for num in range(10)]
```



### Explanation:
- The if/else is part of the **expression**, not the filter  
- This maps every number to a label

---

### Loop vs List Comprehension

### Loop Version:


```python
result = []
for num in range(10):
    result.append(num*2)
result 
```

```python
[num * 2 for num in range(10)]
```



### List Comprehension Version:




### Explanation:

### DICT & SET COMPREHENSIONS 

### Dictionary Comprehensions

#### Example:


```python
grades = {'Ali':88, 'Sara':90, 'Tom':75}
for name,score in grades.items():
    score += 5
    print(score)
```

```python
{name:score+5 for name,score in grades.items()}
```

### Explanation:
- Builds a new dictionary
- Key/value pairs are unpacked using `.items()`
- Adds 5 to each grade

---

### Dictionary from List of Tuples



```python
pairs = [('a',1),('b',2)]
pairs
```

```python
{key: value for key,value in pairs}
```



### Explanation:
- Turns a list of (key, value) pairs into a dictionary
- Very common when working with CSV data or API responses

---

### Set Comprehensions



```python
names = ['Keshvi','David','Daniel']
{name[0] for name in names}
```



### Explanation:
- Curly braces create a **set**, not a list
- Sets remove duplicates automatically
- Useful for extracting unique values

### HIGHER-ORDER FUNCTIONS 

### What Is a Higher-Order Function?

A **higher-order function** is a function that either:

- takes another function as input  
- or returns a function  

In this course, we only need the first part.

Python already uses higher-order functions, for example:
- `sorted(key=...)`
- `map()`
- `filter()`

This is the foundation of functional-style programming.

---

### Example — Passing a Function to Another Function

```python
def apply(func,value):
    return func(value*2)
```

```python
apply(lambda num:num + 10, 30)
```



### Explanation:
- `apply()` takes a function as input  
- It calls that function with the given value  
- The lambda increases the number by 10  
- This shows that functions are *first-class citizens* in Python

---

###  map() and filter() 

### map(): Apply a Function to Each Item

#### Example 1 — Strings to Integers



```python
numbers = ['1','2','3','4','5']
nums = list(map(int,numbers))
nums
```



### Explanation:
- `map(int, numbers)` applies `int()` to each string  
- Converts `"1"` → `1`, `"2"` → `2`, etc.  
- `list()` converts the map object into a list  
- No lambda needed here — we use a built-in function instead

---

### Example 2 — Doubling Numbers



```python
list(map(lambda num:num*2,nums))
```



### Explanation:
- For each `x`, lambda returns `x * 2`
- `map()` is often used for transformations
- Equivalent to:  
  



---

### filter(): Keep Only Matching Items

### Example 1 — Keep Numbers > 10



```python
nums = [24,4,64,5,102,9]

list(filter(lambda num:num>10,nums))
```



### Explanation:
- filter() keeps items where the lambda returns **True**
- For each number, lambda checks the condition  
- Equivalent to:  
  

```python
[num for num in nums if num > 10]
```



---

### Example 2 — Keep Even Numbers



```python
nums = [24,4,64,5,102,9]

list(filter(lambda num:num%2 == 0,nums))
```



### Explanation:
- Same idea: keep items that satisfy the condition  
- Reinforces comprehension logic  
- Ideal for teaching functional filtering
