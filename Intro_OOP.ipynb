## 1️⃣ Motivation — Why a New Way of Thinking?

### 1.1 Real‑World Analogy (Concept Only)
Programming before OOP is like having a toolbox where data and tools are scattered:
- The **data** (e.g., a car’s *speed*, *fuel*) is stored in one place.
- The **functions** (`start_engine`, `accelerate`) live somewhere else.

When you add a new car, you must create a new set of data and remember to call the right functions with the right data. This gets messy fast.

### 1.2 Why This Becomes a Problem
As programs grow:
- Code becomes hard to manage (functions and data drift apart).
- Important data isn’t protected from accidental misuse.
- We repeat ourselves (poor reuse).
- It’s harder to model real‑world things (cars, students, bank accounts).

### 1.3 OOP Idea (Before Any Code)
Bundle **data + behavior** together in a single unit called an **object**.  
Objects are created from **classes**.  
We think in nouns (objects) and attach verbs (behaviors) to them.

## 2️⃣ From Variables → Attributes → Objects

### 2.1 Variables (Refresher)

```python

```

### 2.2 Attributes (Concept First)
An **attribute** is a variable that **belongs to an object**.  
We access it with dot notation: `object.attribute`.

### 2.3 Classes and Objects
- A **class** is a **blueprint** that specifies what data (attributes) and behaviors (methods) its objects have.
- An **object** (or **instance**) is a **concrete thing** created from a class.

## 4️⃣ Adding Attributes with an Initialization Function (Constructor)

### 4.1 What Is Initialization?
When we create an object, we want it to **start with** some values.

### 4.2 What Is the Constructor?
In Python, the **constructor** is the special method named `__init__`.  
It runs **right after** the object is created and sets up its initial state.

### 4.3 What Is `self` (Most Important!)
- `self` means **"this object"**.
- Python passes the object to the method as the first argument.
- When you write `b = Bottle("Water", 500)`, Python effectively calls:  
  `Bottle.__init__(b, "Water", 500)` → so inside `__init__`, `self` is `b`.


````python
class Bottle:
    def __init__(self):
        
        self.label = "Water"
        self.ml = 500
    
````

````python
b1 = Bottle()
````

````python
b1.label
````

````python
b1.ml
````

````python
b2 = Bottle()
b2.label
````

````python
b2.label = "Juice"
b2.label
````

````python
class Bottle:
    def __init__(self,label = 'Nothing',ml = 'N/A'):
        # `self` refers to BRAND NEW object being constructed 
        self.label = label
        self.ml = ml
    
````

````python
b3 = Bottle('G2',750)
b3.label
````

````python
b4 = Bottle('G1',800)
b4.label
b4.ml
````

## 5️⃣ Methods — Adding Behavior to the Noun

A **method** is a function defined **inside a class**. It uses `self` to access or modify **this object’s** attributes.


````python
class Counter:
    def __init__(self, value = 0):
        self.value = value

    def increment(self,step = 1):
        self.value += step #behvaior : change the internal state /`or`/ increase value by step

    def reset(self):
        self.value = 0
        
````

````python
c1 = Counter()
c1.value
````

````python
c1.increment()
c1.value
````

````python
c1.increment(5)
c1.value
````

````python
c1.reset()
c1.value
````

**Why `self` again?**  
It lets the method read and modify **this specific object’s** data (`self.value`).

## 7️⃣ Encapsulation — Protecting Internal State

**Goal:** Keep data safe from accidental misuse; enforce rules in **one place**.

- `_name` (single underscore): **convention** for “internal use” (not enforced).
- `__name` (double underscore): triggers **name mangling** to `_ClassName__name`, discouraging direct access.

We expose **methods** (or **properties**) to control how values change.


````python
class BankAccount:
    def __init__(self,owner,balance = 0):
        self.owner = owner
        self.__balance = balance # 'double underscore' makes it a private attribute
    def deposit(self,amount):
        if amount > 0:
            self.__balance += amount
        else:
            print('Deposit amount must be positive')
    def show_balance(self):
        print(self.owner,"'s balance is:", self.__balance)

    def withdraw(self,amount):
        if amount <= 0: 
            print('Withdraw amount must be positive')
        elif amount <= self.__balance:
            self.__balance -= amount
        else:
            print('Insufficient Funds')
    
        
````

````python
ac1 = BankAccount('Ali')
ac1.owner
````

````python
ac1.deposit(100)
````

````python
ac1.balance
````

````python
ac1.show_balance()
````

````python
ac1.withdraw(80)
````

````python
ac1.show_balance()
````

````python
ac1.withdraw(-50)
````

````python
ac1.withdraw(50)
````

**Why double underscore?**  
Inside `BankAccount`, `__balance` is stored as `_BankAccount__balance`.  
This prevents accidental outside access and keeps changes going through `deposit`/`withdraw`, where we can **validate** them.
