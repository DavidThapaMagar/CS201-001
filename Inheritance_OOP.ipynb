# Inheritance in Object-Oriented Programming 


## 1) Why Inheritance?

Inheritance lets a class (the **child/subclass**) reuse and extend code from another class (the **parent/superclass**). This prevents duplication and naturally models real-world relationships. For example, a `Car` **is a** `Vehicle`. If the `Vehicle` class already defines behaviors like `info()`, then `Car` can automatically use them without rewriting code.

```python
class Vehicle:
    def info(self):
        return "Generic Vehicle"
        
class Car(Vehicle):
    pass
```

```python
v = Vehicle()
c = Car()
```

```python
v.info()
```

```python
c.info()
```

## 2) Overriding vs. Extending Parent Methods

**Goal:** Learn the crucial difference between *overriding* (replace the parent behavior) and *extending* (keep the parent behavior and add to it using `super()`).

### When to Override
Override when the subclass’s behavior should be **different** and you don’t need the parent logic.

### When to Extend
Extend when the parent method already performs useful work you still need (e.g., setup or validation). In this case, call `super().method_name()` and then add more behavior.

### 2.1 Overriding Example

```python
class Vehicle:
    def info(self):
        return "Generic Vehicle"
        
class Car(Vehicle):
    def info(self):
        return "Car starting with key ignition..."
```

```python
v = Vehicle()
c = Car()
```

```python
v.info()
```

```python
c.info()
```

### 2.2 Extending Example (override + `super()`)
Here we keep the parent’s behavior and append details specific to the subclass.

```python
class Vehicle:
    def start(self):
        return "Vehicle starting"
        
class Car(Vehicle):
    def start(self):
        base = super().start()
        return base + " in silent mode."
```

```python
v = Vehicle()
c = Car()
```

```python
v.start()
```

```python
c.start()
```

### 2.3 Extending `__init__` with `super()`
If the parent sets important state in `__init__`, the child must call `super().__init__(...)` so that setup still happens.

```python
class Vehicle:
    def __init__(self,brand):
        self.brand = brand

    def info(self):
        return self.brand
        
class Car(Vehicle):
    def __init__(self,brand,model):
        super().__init__(brand)
        self.model = model

    def info(self):
        return super().info()+ " " + self.model
```

```python
v = Vehicle('Toyota')
c = Car('Honda', 'Civic')
```

```python
v.info()
```

```python
c.info()
```

## 3) Special (Dunder) Methods We’ll Use: `__str__`, `__repr__`, `__len__`, `__contains__`, `__iter__`

**Goal:** Make our classes feel like built-in Python objects.

- **`__str__`**: human‑friendly display for `print(obj)`.
- **`__repr__`**: developer‑friendly representation (good for debugging/logging).
- **`__len__`**: number of contained items (`len(obj)`).
- **`__contains__`**: membership test (`x in obj`).
- **`__iter__`**: iteration support (`for x in obj:`).

### 3.2 Container Behavior: `len`, `in`, and Iteration
We’ll model a small `Fleet` that holds vehicles. These dunders let it behave naturally as a container.

```python
class Fleet:
    def __init__(self):
        self.vehicles = []
    
    def add(self,v):
        return self.vehicles.append(v)
    
    def __len__(self): # to find length
        return len(self.vehicles)

    def __iter__(self): # to iterate through the objects
        return iter(self.vehicles)

    def __contains__(self,v):
        return v in self.vehicles
```

```python
for current_object in fleet:
    print(current_object)
```

## 4) Protected and Private Attributes in Inheritance

**Goal:** Use naming conventions to communicate which attributes are intended for internal use.

Python doesn’t enforce access modifiers, but we follow conventions:
- **`_name`**: *protected by convention* → internal use; subclasses may access but should be careful.
- **`__name`**: *private-ish via name‑mangling* → discourages direct access; prefer methods/properties to read it.

```python
class Vehicle:
    def __init__(self,brand,vin):
        self._brand = brand
        self.__vin = vin # private attribute

    def get_vin(self):
        return self.__vin

class Car(Vehicle):
    def get_vin(self):
        return self._brand + " " + str(super().get_vin())
        
        
        
```

```python
v = Vehicle('Toyota', 123)
c = Car('Honda', 456)
```

```python
v.get_vin()
```

```python
c.get_vin()
```

## 5) `@property` — What It Is and Why It Helps

**Goal:** Provide attribute-style access while keeping encapsulation.

`@property` turns a method into a read‑only attribute. It gives clean syntax today and flexibility tomorrow (you can add validation or compute values later without changing how callers access it).

### Without `@property`

```python
class Car:
    def __init__(self,speed):
        self.speed = speed

    def get_speed(self):
        return self.speed
```

```python
c = Car(60)
c.get_speed()
```

### With `@property`

```python
class Car:
    def __init__(self,speed):
        self.speed = speed
    @property
    def get_speed(self):
        return self.speed
```

```python
c = Car(60)
c.get_speed
```

## 6) Composition vs. Inheritance

**Goal:** Choose the right relationship.

- **Inheritance** expresses an **is‑a** relationship: `Car` is a `Vehicle`.
- **Composition** expresses a **has‑a** relationship: `Car` has an `Engine`.

Prefer composition when you’re adding capabilities that don’t fundamentally change *what the object is*. It keeps designs flexible and avoids deep hierarchies.

### Example 1: Car **has an** Engine (Composition)

```python
class Engine:
    def __init__(self,hp,size):
        self.hp = hp
        self.size = size

    def __str__(self):
        return str(self.hp)+ " " + str(self.size)
```

```python
class Vehicle:
    def __init__(self,brand):
        self.brand = brand

    def __str__(self):
        return self.brand

class Car(Vehicle):
    def __init__(self,brand,model,engine):
        super().__init__(brand)
        self.model = model
        self.engine = engine

    def __str__(self):
        return super().__str__() + " " + self.model + " " + str(self.engine)
```

```python
eng1 = Engine(200,2.5)
car1 = Car('Honda','Civic',eng1)
```

```python
print(car1)
```

### Example 2: Garage **has** Cars (Composition)

```python
class Garage:
    cars = []
    def __init__(self):
        self.cars = []

    def add(self,car):
        self.cars.append(car)

    def list_cars(self):
        print(self.cars)
```

```python
eng2 = Engine(350,2.5)
car2 = Car('Toyota','4Runner',eng2)
```

```python
garage = Garage()
garage.add(car1)
```

```python
garage.add(car2)
```

## 7) Multilevel Inheritance (Clean Chain)

**Goal:** See how a deeper chain works in practice.

We’ll build a simple chain `ElectricCar → Car → Vehicle` and override `info()` at each level.

```python
class Vehicle:
    kind = "Generic"
    def __init__(self,brand,vin):
        self.brand = brand 
        self.vin = vin 

    def info(self):
        return self.brand + " " + str(self.vin)
```

```python
class Car(Vehicle):
    kind = "Car"
    def __init__(self,brand,model,vin):
        super().__init__(brand,vin)
        self.model = model

    def info(self):
        return super().info() + " " + self.model
```

```python
class ElectricCar(Car):
    kind = "ElectricCar"
    def __init__(self,brand,model,vin,battery_mph):
        super().__init__(brand,model,vin)
        self.battery_mph = battery_mph

    def info(self):
        return super().info() + " " + str(self.battery_mph)
```

**Goal:** Combine inheritance, composition, encapsulation, and the special methods we kept (`__str__`, container dunders).

We’ll define `Vehicle → Car → ElectricCar`, a `ServiceRecord` helper, and a `Fleet` container.

Step 1 – Service Records (Composition Helper)

Create a class ServiceRecord with:

Step 2 – Base Class with Encapsulation

Define a base class Vehicle that includes:

    - stores a private __vin.
- A read-only @property vin.
- A method add_service(self, note) that appends a new ServiceRecord.
- An info(self) method returning "<kind> <brand> | VIN <vin>".
- A __str__ method returning the same string as info().

Step 3 – First Subclass (Car)

Create a class Car(Vehicle) that:
- Sets kind = "Car".
- Has its own initializer adding a model attribute, while calling super().__init__.
- Overrides info() to append " | Model: <model>".
    
Step 4 – Multilevel Subclass (Electric Car)

Create a class ElectricCar(Car) that:
- Sets kind = "Electric Car".
- Adds an attribute battery_kwh.
- Overrides info() to append " | Battery: <battery_kwh>kWh".
    
Step 5 – Fleet Container (Composition + Dunder Methods)

Design a Fleet class that:
- Stores a name and a list vehicles.
- Provides add_vehicle(self, vehicle) to append to the list.
- Implements the following dunder methods:
    - __len__ → returns number of vehicles.
    - __contains__ → enables v in fleet.
    - __iter__ → enables iteration over vehicles.
    - __str__ → returns "Fleet <name>: <veh1>, <veh2>, ..." using each vehicle’s string.
    
Step 6 – Final Assembly

Use your classes together to:
- Create a Fleet("Campus Vehicles").
- Add at least one Car and one ElectricCar.
- Record at least one service note for each vehicle.
- Display:
    - The number of vehicles in the fleet (len(fleet)).
    - A single-line summary of the fleet (print(fleet)).
    - Each vehicle on its own line via a for loop.

```python
from datetime import date

class ServiceRecord:
    def __init__(self, when, note):
        self.when = when
        self.note = note
    def __repr__(self):
        return f"ServiceRecord({self.when}, {self.note!r})"

class Vehicle:
    kind = "Generic"
    def __init__(self, brand, vin):
        self.brand = brand
        self._service = []       # protected
        self.__vin = vin         # private-ish
    @property
    def vin(self):
        return self.__vin
    def add_service(self, note):
        self._service.append(ServiceRecord(date.today(), note))
    def info(self):
        return f"{self.kind} {self.brand} | VIN {self.vin}"
    def __str__(self):
        return self.info()

class Car(Vehicle):
    kind = "Car"
    def __init__(self, brand, model, vin):
        super().__init__(brand, vin)
        self.model = model
    def info(self):
        return super().info() + f" | Model: {self.model}"

class ElectricCar(Car):
    kind = "Electric Car"
    def __init__(self, brand, model, battery_kwh, vin):
        super().__init__(brand, model, vin)
        self.battery_kwh = battery_kwh
    def info(self):
        return super().info() + f" | Battery: {self.battery_kwh}kWh"

class Fleet:
    def __init__(self, name):
        self.name = name
        self.vehicles = []
    def add_vehicle(self, vehicle):
        self.vehicles.append(vehicle)
    def __len__(self):
        return len(self.vehicles)
    def __contains__(self, v):
        return v in self.vehicles
    def __iter__(self):
        return iter(self.vehicles)
    def __str__(self):
        return f"Fleet {self.name}: " + ", ".join([str(v) for v in self.vehicles])

f = Fleet("Campus Vehicles")
c1 = Car("Toyota", "Corolla", "T001")
e1 = ElectricCar("Tesla", "Model 3", 60, "E100")

f.add_vehicle(c1)
f.add_vehicle(e1)

c1.add_service("Oil change")
e1.add_service("Battery check")

print(len(f))
print(str(f))
for v in f:
    print(v)
```
