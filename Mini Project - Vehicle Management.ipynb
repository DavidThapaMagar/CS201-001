8) Mini Project — Vehicle Management

Goal: Combine inheritance, composition, encapsulation, and the special methods we kept (__str__, container dunders).

We’ll define Vehicle → Car → ElectricCar, a ServiceRecord helper, and a Fleet container.

Step 1 – Service Records (Composition Helper)

Create a class ServiceRecord with:
- Attributes: when (a date object) and note (string).
- A developer-friendly __repr__ that shows the date and note clearly.

Step 2 – Base Class with Encapsulation

Define a base class Vehicle that includes:
- A class attribute kind = "Generic".
- An initializer __init__(self, brand, vin) that
    - stores brand (public),
    - creates a protected list _service = [],
    - stores a private __vin.
- A read-only @property vin.
- A method add_service(self, note) that appends a new ServiceRecord.
- An info(self) method returning "<kind> <brand> | VIN <vin>".
- A __str__ method returning the same string as info().

Step 3 – First Subclass (Car)

Create a class Car(Vehicle) that:
- Sets kind = "Car".
- Has its own initializer adding a model attribute, while calling super().__init__.
- Overrides info() to append " | Model: <model>".
    
Step 4 – Multilevel Subclass (Electric Car)

Create a class ElectricCar(Car) that:
- Sets kind = "Electric Car".
- Adds an attribute battery_kwh.
- Overrides info() to append " | Battery: <battery_kwh>kWh".
    
Step 5 – Fleet Container (Composition + Dunder Methods)

Design a Fleet class that:
- Stores a name and a list vehicles.
- Provides add_vehicle(self, vehicle) to append to the list.
- Implements the following dunder methods:
    - __len__ → returns number of vehicles.
    - __contains__ → enables v in fleet.
    - __iter__ → enables iteration over vehicles.
    - __str__ → returns "Fleet <name>: <veh1>, <veh2>, ..." using each vehicle’s string.
    
Step 6 – Final Assembly

Use your classes together to:
- Create a Fleet("Campus Vehicles").
- Add at least one Car and one ElectricCar.
- Record at least one service note for each vehicle.
- Display:
    - The number of vehicles in the fleet (len(fleet)).
    - A single-line summary of the fleet (print(fleet)).
    - Each vehicle on its own line via a for loop.

```python
from datetime import date
class ServiceRecord:
    def __init__(self, when, note):
        self.when = when
        self.note = note
    def __repr__(self):
        return f"ServiceRecord({self.when}, {self.note!r})"

class Vehicle:
    kind = "Generic"
    def __init__(self,brand,vin):
        self.brand = brand 
        self.__vin = vin
        self._service = []

    @property
    def vin(self):
        return self.__vin

    def add_service(self, note):
        self._service.append(ServiceRecord(date.today(), note))

    def info(self):
        return f"{self.kind} {self.brand} | VIN {self.vin}"

    def __str__(self):
        return self.info()

class Car(Vehicle):
    kind = "Car"
    def __init__(self,brand,model,vin):
        super().__init__(brand,vin)
        self.model = model

    def info(self):
        return super().info() + f" | Model: {self.model}"

class ElectricCar(Car):
    kind = "Electric Car"
    def __init__(self,brand,model,vin,battery_kwh):
        super().__init__(brand,model,vin)
        self.battery_kwh = battery_kwh

    def info(self):
        return super().info() + f" | Battery: {self.battery_kwh}kwh"

class Fleet:
    def __init__(self,name):
        self.name = name
        self.vehicles = []

    def add_vehicle(self,vehicle):
        self.vehicles.append(vehicle)
    def __len__(self):
        return len(self.vehicles)
    def __contains__(self,a):
        return a in self.vehicles
    def __iter__(self):
        return iter(self.vehicles)
    def __str__(self):
        return f"Fleet {self.name}: " + ", ".join([str(v) for v in self.vehicles])
        
```

```python
f = Fleet("Campus Vehicles")
c1 = Car("Toyota", "Corolla", "T001")
e1 = ElectricCar("Tesla", "Model 3", 60, "E100")

f.add_vehicle(c1)
f.add_vehicle(e1)

c1.add_service("Oil change")
e1.add_service("Battery check")

print(len(f))
print(str(f))
for v in f:
    print(v)
```
