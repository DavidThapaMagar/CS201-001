```python
def traverse_sim(head_index,storage):
    output = []
    index = head_index
    while index is not None:
        value, index = storage[index]
        output.append(value)
    return output
        
```

```python
traverse_sim(0,[('A',1),('B',2),('C',None),])
```

```python
# adding a new node in the end 

def append_sim(value,head_index,storage):
    new_index = len(storage)
    storage.append((value,None))
    if head_index is None:
        return new_index, storage
    index = head_index
    while True:
        value,next_index = storage[index]
        if next_index is None:
            storage[index] = (value, new_index)
            break
        index = next_index 
    return head_index, storage
```

```python
linked_list = [(20,1),(40,None)]
append_sim(60, 0, linked_list)
```

```python
def groupAnagrams(strs):
    container = []
    for char in strs:
        anagram = ''.join(sorted(char))
        container[anagram].append(char)
    print(container)
```

```python
strs = ['cat','mat','sat']
groupAnagrams(strs)
```
