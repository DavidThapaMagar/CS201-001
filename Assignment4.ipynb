#### Question 1 (20 percent)

Design a small object-oriented program for managing book information in a personal library system.

Create a class that represents a Book, where each book has essential information such as:
- a title
- an author
- a publication year
- at least one additional attribute of your choice (for example: genre, ISBN, number of pages, and so on)

Your tasks:
1. Define the Book class with appropriate attributes, initialized in the constructor.
2. Add at least two meaningful instance methods (for example: updating information, checking a condition, returning a formatted description, and so on).
3. Demonstrate that the class works by creating multiple Book objects and using your methods on them (for example, print their information, update something, check something).

Write the class and the test code from scratch.


````python
# Question 1 solution
class Books:
    def __init__(self,title,author,year,isbn):
        self.title = title
        self.author = author
        self.year = year
        self.__isbn = isbn

    def update_title(self,title):
        self.title = title

    def update_author(self,author):
        self.author = author

    def update_year(self,year):
        self.year = year
    
    def update_isbn(self,isbn):
        self.__isbn = isbn
    
    def info(self):
        return f"ISBN: {self.__isbn} | Title: {self.title} | Author: {self.author} | Year: {self.year}"

    def __str__(self):
        return info
````

````python
b1 = Books('Harry Potter','JK Rowlings',2008,1239418)
b1.info()
````

````python
b1.update_year(2009)
b1.info()
````

````python
b1.update_author('John Stones')
b1.info()
````

You are asked to build a small system to represent different types of employees in a company.

Your tasks:

   - Create a parent class that represents a general employee (for example, with attributes such as name, employee ID, and any other attribute you choose).

2. Child classes with polymorphism
   - Create at least two subclasses that inherit from the parent class (for example, FullTimeEmployee, PartTimeEmployee, Intern, and so on).
   - Each subclass must add its own unique attributes and override at least one method from the parent class to demonstrate polymorphism (the overridden method can be __str__ or another method).

3. Demonstration of polymorphism
   - Create several objects from different subclasses.
   - Store them in a single list and loop over that list, calling the same method on each object (for example, printing them or calling a method that you defined).
   - Show that the behavior depends on the actual class of each object (this is polymorphism).

4. Short written explanation
   - In 2 to 4 sentences, explain why printing a child class object works even if the __str__ special method is defined in the parent class.
   - In your explanation, refer to inheritance and how Python looks up methods.


````python
# Question 2 solution
class Employee:
    def __init__(self,id,name,position,phno):
        self.name = name
        self.id = id
        self.position = position
        self.phno = phno
 
    def __str__(self):
        return f"ID: {self.id} | Name: {self.name} | Position: {self.position} | Phone Number: {self.phno}"
        
class FullTimeEmployee(Employee):
    def __init__(self,id,name,position,phno,salary,status = "Full Time Employee"):
        super().__init__(id,name,position,phno)
        self.status = status
        self.salary = salary

    def __str__(self):
        super.str
        return f"ID: {self.id} | Name: {self.name} | Position: {self.position} | Phone Number: {self.phno} | Status: {self.status} | Salary: {self.salary}$"

class Intern(Employee):
    def __init__(self,id,name,position,phno,school,status = "Intern"):
        super().__init__(id,name,position,phno)
        self.status = status
        self.school = school
        
    def __str__(self):
        return f"ID: {self.id} | Name: {self.name} | Position: {self.position} | Phone Number: {self.phno} | Status: {self.status} | School: {self.school}"
````

````python
e1 = FullTimeEmployee(123,'David','CEO',8122239121,21000)
print(e1)
````

````python
e2 = Intern(124,'Ryan','JrSWE',121251212,'Indiana State University')
print(e2)
````

Use the same orders list from Question 3:

Your tasks:
1. Filtering with filter and lambda
   - Use filter together with a lambda function to obtain a list of all orders that are from Canada and have a total of at least 100.

2. Mapping with map and lambda
   - Use map with a lambda function to create a list that contains only the customer names from all orders.

3. Your own higher order function
   - Define a function named transform_totals(orders, func) that:
     - takes the list orders and a function func as parameters
     - applies func to the total value of each order
     - returns a new list containing the transformed totals (one per order, in order).
   - Then:
     - call transform_totals with a lambda that adds 15 to each total and store or print the result
     - call transform_totals again with a lambda that applies a 20 percent tax (multiply by 1.2) and store or print the result.


````python
# Question 4 solution
orders = [
    {"id": 1, "customer": "Alex", "total": 120.0, "status": "shipped", "country":"Canada"},
    {"id": 2, "customer": "Maria", "total": 75.5, "status": "pending", "country": "USA"},
    {"id": 3, "customer": "John", "total": 200.0, "status": "shipped", "country": "Canada"},
    {"id": 4, "customer": "Sara", "total": 50.0, "status": "cancelled", "country": "Mexico"},
    {"id": 5, "customer": "Lin", "total": 300.0, "status": "pending", "country": "Canada"},
]
````

````python
canada = list(filter(lambda order: order['country'] == 'Canada' and order['total'] >= 100, orders))
````

````python
print(canada)
````

````python
name = list(map(lambda order: order['customer'],orders))
print(name)
````

````python
total_add15 = lambda total: total + 15

def transform_totals(orders, func):
    return [func(order['total']) for order in orders]

transform_totals(orders, total_add15)
````

````python
transform_totals(orders, lambda total:total*1.2)
````
