```python
def dec2bin(num):
    stack = []
    while num > 0:
        stack.append(num % 2)
        num  = num // 2 
    binary = ''
    while stack:
        binary += str(stack.pop())
    return int(binary)
```

```python
def append_sim(value, head_index, storage):
    new_idx = len(storage)
    storage.append((value,None))
    if head_index is None:
        return new_idx, storage
    idx = head_index
    while True: 
        val,nex_idx = storage[idx]
        if next_idx is None:
            storage[idx] = (val,new_idx)
            break
        idx = next_idx
    return head_index, storage
```

```python
def subset(set1,set2): 
    if set1 != set2: 
        intersection_set = set1.intersection(set2)
        set1only = set1 - intersection_set
        set2only = set2 - intersection_set
        print(set1only.union(set2only))
    else: 
        print('Both Sets have same values.')
```
