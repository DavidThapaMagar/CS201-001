#### `range` function

The range object is:
 - suscriptable
 - iterable
 - immutable
 - ordered

However, you will generally see it used for loops, so the iterable property is the only one typcially seen. 

```python
# basic use of range
for i in range (5):
    print(i)
# create a list from a range object
```

```python
# change start index
for i in range (1,6):
    print(i)
```

```python
# change step size
for i in range (1,6,2):
    print(i)
```

### Mapping types: `dict`

#### Dictionary 

Lists may be constructed in several ways:

 - using `key:value` pairs with braces:
    - empty list: `{}`
    - separating items with commas: `{'a':1, 'b':2, 'c':3}`
 - dictionary comprehension: 
    - `{x:x**2 for x in range(10)}`
 - type constructor: 
    - empty list: `dict()` 
    - from iterable: `dict([('a', 1), ('b', 2), ('c', 3)])`, `dict(a=1, b=2, c=3)`

```python
# create some dictionaries
dic1 = {}
dic1
```

```python
dic2 = {'Name':'David','Major':'Computer Science','Grade':'85'}
dic2
```

#### Dictionary operations

Dictionaries various [methods/operations](https://docs.python.org/3/library/stdtypes.html#dict) as noted in the docs.

##### Dictionary  properties

The important properties of dictionaries are as follows:
 - access values by keys
 - iterable
 - mutable
 - unordered
 - keys must be hashable
 - can be nested
 - dynamic


```python
# access via key, keys, values, items, iterate, mutable: d[k]=v, del, update, ordered, arbitrary objects, nesting


# access via key
dic2['Name']
```

```python
# keys
dic2.keys()
```

```python
# values
dic2.values()
```

```python
# items
dic2.items()
```

```python
# iterate over keys

for key in dic2.keys():
    print(key)
```

```python
# iterate over values

for values in dic2.values():
    print(values)
```

```python
# iterate over key-value pairs

for key,values in dic2.items():
    print(key,values)
```

```python
# mutable

dic2['Grade'] = 88
dic2
```

```python
# del
del dic2['Major']
dic2 
```

```python
# update
dic2.update({'Age':22,'City':'Terre Haute'})
dic2
```

```python
# ordered
sorted(dic2)
```

```python
# arbitrary objects
dic2['Address'] = [101,'University','St',48123]
dic2
```

```python
# nesting
students = {
    'Name':['Ali','Bob','Rose'],
    'Grade':[80,70,80]
}
students
```

```python
nested_dic = {'dic1':students,'dic2':dic2}
nested_dic
```

```python
for key,value in nested_dic.items():
    print('main keys:', key)
    for in_key,in_value in nested_dic[key].items():
        print('sub keys:', in_key)
        print('in values: ',in_value)
```
