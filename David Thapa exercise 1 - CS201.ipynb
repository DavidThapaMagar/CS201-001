## Conditional Logic

- Ask for temperature in Celsius.

````python
temp = int(input('Enter Temperature in C: '))
if temp > 30:
    print('Its hot!')
elif temp >= 15 and temp <=30:
    print('Nice weather')
else:
    print("It's cold!")
````

## Set Practice

- Create a set of 5 favorite fruits.
- Add one, remove one.
- Make a second set and find union, intersection, difference.

````python
fruits = {'Apple', 'Banana', 'Mango', 'Pineapple', 'Grapes'}
fruits.add('Papaya')
fruits
````

````python
fruits.remove('Apple')
fruits
````

````python
fruits2 = {'Grapes', 'Dragon Fruit', 'Banana', 'Litchi'}
union = fruits | fruits2
print(union)
````
